### business problem
_
specific problem business wants to solve?
- Goal is to predict if the client will subscribe a term deposit.
- Most likely Bank needs to reduce spending on phone calling campaign by targeting classified clients.
- Right now bank calls random clients -> more recall, less precise
- the price of a mistake is unknown
- to evaluate success of the solution implementation, bank needs to divide clients in two groups: in one group choose clients using model, randomly - in another. Having same phone calls budget compare conversion and number of subscriptions


### ML problem
Binary classification
Target: y = 1 if the client subscribed a term deposit at the end of the marketing campaign; y = 0 if the client didn't subscribe a term deposit after a phone call.
prediction time: before the phone call.


### Metrics
Main metric: Precision, because important not to make unsuccessful phone calls to reduce spending.
Additionally: PR AUC - to check quality of the prediction.

### Risks and Assumptions
- Check if the features available at the prediction moment
- New clients don't have previous phone calling campaigns info
